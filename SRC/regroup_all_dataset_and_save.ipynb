```python
import json
from pathlib import Path
from typing import Dict, Any
from datetime import date
from  API.preprocessing.get_historical_stock_preprocessing import historical_stock_info
from API.preprocessing.get_company_news_preprocessing import create_news_dataset
from  API.preprocessing.get_news_country_preprocessing import get_news_country
from  API.preprocessing.get_info_company_preprocessing import get_info_company_preprocessing
from  API.preprocessing.get_options_preprocessing import create_options_dataset
```

```python
def save_json(data: Dict[str, Any], filepath: str) -> None:
    """
    Save an dict object in an json file

    Args:
        data (dict): the data to save 
        filepath (str): the path in which the json file is created
    """
    path = Path(filepath)
    try:
        with path.open("w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
        print(f"Data save at : {path.resolve()}")
    except Exception as e:
        print(f"Error whan saving the file : {e}")
```

```python
def regroupe_all_dataset(business_ticker, business_name):
    info_company = get_info_company_preprocessing(business_ticker)
    company_news = create_news_dataset(business_name=business_name, business_ticker=business_ticker)
    country_news = get_news_country()
    historical_stock = historical_stock_info(business_ticker)
    options = create_options_dataset(business_ticker)
    today = date.today().strftime("%Y_%m_%d")
    data = {
        "info_company":info_company,
        "news": {
            "company_news":company_news,
            "country_news":country_news
        },
        "historical_stock_info":historical_stock,
        "options":options
    }
    filepath = rf'..\data\output\all_data_regroup\{business_ticker.lower()}_data_{today}.json'
    save_json(data=data, filepath=filepath)
    return data
```
